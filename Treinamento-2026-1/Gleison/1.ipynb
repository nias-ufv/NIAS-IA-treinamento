```python
tweets = [
"Wow, what a great day today!! #sunshine",
"I feel sad about the things going on around us. #covid19",
"I'm really excited to learn Python with @JovianML #zerotopandas",
"This is a really nice song. #linkinpark",
"The Python programming language is useful for data science",
"Why do bad things happen to me?",
"Apple announces the release of the new iPhone 12. Fans are excited.",
"Spent my day with family!! #happy",
"Check out my blog post on common string operations in Python. #zerotopandas",
"Freecodecamp has great coding tutorials. #skillup"
]
```

```python
happy_words = ['great', 'excited', 'happy', 'nice', 'wonderful',
'amazing', 'good', 'best']

sad_words = ['sad', 'bad', 'tragic', 'unhappy', 'worst']
```

```python
is_happy =False

for tweet in tweets:
    for word in happy_words:
        if word in tweet:
            is_happy = True
            break

print(is_happy)
```

```python
happys ={}
sads ={}

for i in range(len(tweets)):
  x = True
  if  x:
        for word in happy_words:
            if word in tweets[i].strip("#,.;?!"):
                happys[i] = word
                x=False
                break


        for word in sad_words:
                if word in tweets[i].strip("#,.;?!"):
                    sads[i] = word
                    x=False
                    break        
    



print(f'Existem {len(happys)} tweets felizes,que são:')
for elemento in happys:
     
     print(f"A frase '{tweets[elemento]}' é feliz pois tem a palavra {happys[elemento]}")
     
print()

print(f'Existem {len(sads)} tweets tristes,que são:')
for elemento in sads:
     
     print(f"A frase '{tweets[elemento]}' é triste pois tem a palavra {sads[elemento]}")
    
```
